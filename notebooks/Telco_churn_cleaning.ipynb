```python

import pandas as pd
import numpy as np

df = pd.read_csv('telco_churn.csv')

df.head()
```

```python
# How many rows and columns?
print("Shape:", df.shape)


# See all column names
print("\nColumns:")
print(df.columns.tolist())

# See data types of each column
print("\nData types:")
print(df.dtypes)
```

```python
# Full dataset summary — non-null counts + data types
df.info()

# Statistical summary of numeric columns
df.describe()
```

```python
# Check for actual NaN / None values
print("Missing values per column:")
print(df.isnull().sum())

# TotalCharges has BLANK STRINGS (not NaN) — find them
blank_mask = df['TotalCharges'].str.strip() == ''
print(f"\nBlank TotalCharges rows: {blank_mask.sum()}")

# Look at those rows
df[blank_mask][['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']]
```

```python
# Check unique values in key categorical columns
cat_cols = ['Churn', 'Contract', 'InternetService',
            'PaymentMethod', 'gender']

for col in cat_cols:
    print(f"\n{col}: {df[col].unique()}")
    print(f"  Value counts:\n{df[col].value_counts()}")
```

```python
# Step 1: Replace blank strings with NaN so pandas can handle them
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)

# Step 2: Convert the column from text (object) to float (decimal number)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Step 3: Fill the 11 NaN values with 0
# (new customers with tenure=0 have been charged nothing yet)
df['TotalCharges'] = df['TotalCharges'].fillna(0)

# Verify the fix
print(df['TotalCharges'].dtype)           # Should now say: float64
print(df['TotalCharges'].isnull().sum()) # Should say: 0
print(df['TotalCharges'].describe())      # Should show min=0, max~8685
```

```python
# Keep original column for reference
df['Churn_label'] = df['Churn']

# Convert: Yes → 1, No → 0
df['Churn'] = (df['Churn'] == 'Yes').astype(int)

# Verify
print(df['Churn'].value_counts())
# 0 = 5174 (stayed),  1 = 1869 (left)

# Instant churn rate — no CASE WHEN needed!
churn_rate = df['Churn'].mean() * 100
print(f"\nOverall churn rate: {churn_rate:.2f}%")
```

```python
# Save cleaned version — always do this after cleaning
df.to_csv('churn_cleaned.csv', index=False)

print("Saved! File: churn_cleaned.csv")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# Final check — no nulls anywhere
print("\nNull check (should all be 0):")
print(df.isnull().sum().sum(), "total nulls")
```

```python
# Detailed stats using numpy for tenure
tenure = df['tenure'].values  # convert to numpy array

print("--- Tenure (months) ---")
print(f"Mean:   {np.mean(tenure):.1f}")
print(f"Median: {np.median(tenure):.1f}")
print(f"Std:    {np.std(tenure):.1f}")
print(f"Min:    {np.min(tenure)}")
print(f"Max:    {np.max(tenure)}")

# Same for MonthlyCharges
monthly = df['MonthlyCharges'].values

print("\n--- Monthly Charges ($) ---")
print(f"Mean:   {np.mean(monthly):.2f}")
print(f"Median: {np.median(monthly):.2f}")
print(f"Std:    {np.std(monthly):.2f}")

# Percentiles — where do most customers fall?
print("\n--- Tenure percentiles ---")
for p in [25, 50, 75, 90]:
    print(f"  {p}th percentile: {np.percentile(tenure, p):.0f} months")
```

```python
# Split into two groups
churned  = df[df['Churn'] == 1]
retained = df[df['Churn'] == 0]

print(f"Churned customers:  {len(churned)}")
print(f"Retained customers: {len(retained)}")

# Compare average tenure
print("\n--- Average tenure (months) ---")
print(f"  Churned:  {churned['tenure'].mean():.1f}")
print(f"  Retained: {retained['tenure'].mean():.1f}")

# Compare average monthly charges
print("\n--- Average monthly charges ($) ---")
print(f"  Churned:  {churned['MonthlyCharges'].mean():.2f}")
print(f"  Retained: {retained['MonthlyCharges'].mean():.2f}")

# Median too (less affected by outliers)
print("\n--- Median monthly charges ($) ---")
print(f"  Churned:  {np.median(churned['MonthlyCharges']):.2f}")
print(f"  Retained: {np.median(retained['MonthlyCharges']):.2f}")
```

Insights - churned customers have shorter average tenure (18 months vs 38) AND higher monthly charges `($74 vs $61)`. They're new AND expensive — the riskiest combination.

```python
# Churn rate by contract type
contract_churn = df.groupby('Contract')['Churn'].agg(
    total='count',
    churned='sum',
    churn_rate='mean'
).round(3)

print("Churn by Contract:")
print(contract_churn.sort_values('churn_rate', ascending=False))

# Churn rate by internet service
internet_churn = df.groupby('InternetService')['Churn'].mean().round(3) * 100
print("\nChurn rate (%) by Internet Service:")
print(internet_churn.sort_values(ascending=False))

# Churn rate by payment method
payment_churn = df.groupby('PaymentMethod')['Churn'].mean().round(3) * 100
print("\nChurn rate (%) by Payment Method:")
print(payment_churn.sort_values(ascending=False))
```

Insights -  It coincides with the SQL findings — this validates both analyses. Month-to-month ~42%, Fiber optic ~42%, Electronic check ~45%. Python got there in 15 lines vs 30+ lines of SQL.

```python
# Correlation of numeric columns with Churn
# Only works with numbers, so select numeric columns only
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges',
                'SeniorCitizen', 'Churn']

corr = df[numeric_cols].corr()['Churn'].drop('Churn')

print("Correlation with Churn (sorted):")
print(corr.sort_values(ascending=False).round(3))

# How to read this:
# Positive = higher value → more churn
# Negative = higher value → less churn
print("\nInterpretation:")
print("  tenure          → negative = longer tenure, less churn (good)")
print("  MonthlyCharges  → positive = higher bill, more churn (risky)")
print("  SeniorCitizen   → positive = seniors churn more")
```

Insights - Tenure is the single strongest numeric predictor of churn — the longer someone stays, the less likely they are to leave.
